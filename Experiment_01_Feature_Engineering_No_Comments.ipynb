```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
cancer = load_breast_cancer()
df = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df['diagnosis'] = cancer.target
df['diagnosis'] = df['diagnosis'].map({0: 'Malignant', 1: 'Benign'})
print('Original Dataset')
display(df.head())
print('Dataset Shape:', df.shape)
```

```python
print('Missing Values in Each Column:')
print(df.isnull().sum())
print('Total Missing Values:', df.isnull().sum().sum())
```

```python
df.loc[0:4, 'mean radius'] = np.nan
df.loc[5:9, 'mean texture'] = np.nan
print('Missing Values After Introduction:')
print(df.isnull().sum().sum())
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns
imputer = SimpleImputer(strategy='mean')
df[numeric_columns] = imputer.fit_transform(df[numeric_columns])
print('Missing Values After Handling:')
print(df.isnull().sum().sum())
display(df.head())
```

```python
encoder = LabelEncoder()
df['diagnosis'] = encoder.fit_transform(df['diagnosis'])
print('After Label Encoding:')
display(df[['diagnosis']].head())
print('Encoded Classes:', encoder.classes_)
```

```python
scaler = StandardScaler()
df[numeric_columns] = scaler.fit_transform(df[numeric_columns])
print('After Feature Scaling:')
display(df.head())
```

```python
print('Final Model-Ready Dataset')
display(df.head(10))
print('Dataset Shape:', df.shape)
print('Total Missing Values:', df.isnull().sum().sum())
print('Statistical Summary:')
display(df.describe())
```
