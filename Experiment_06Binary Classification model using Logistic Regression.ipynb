```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, confusion_matrix, ConfusionMatrixDisplay,
    precision_score, recall_score, f1_score, classification_report
)
```

```python
df = pd.read_csv("data.csv")
display(df.head())
print("Shape:", df.shape)
print("\nMissing Values")
print(df.isnull().sum())
print("\nInformation")
df.info()
print("\nStatistical Summary")
display(df.describe())
```

```python
df = df.drop(["id", "Unnamed: 32"], axis=1, errors="ignore")
df["diagnosis"] = df["diagnosis"].map({"M": 1, "B": 0})
X = df.drop("diagnosis", axis=1)
y = df["diagnosis"]
print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("Target values:")
print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
print("Training and testing features standardized.")
```

```python
model = LogisticRegression(random_state=42, max_iter=1000)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print("Model training and prediction completed.")
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", round(accuracy, 4))
```

```python
cm = confusion_matrix(y_test, y_pred)
print(cm)
display_plot = ConfusionMatrixDisplay(confusion_matrix=cm)
display_plot.plot(cmap="Blues")
plt.title("Confusion Matrix")
plt.show()
```

```python
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
results = pd.DataFrame({
    "Metric": ["Accuracy", "Precision", "Recall", "F1 Score"],
    "Value": [accuracy, precision, recall, f1]
})
display(results)
```

```python
print(classification_report(y_test, y_pred, target_names=["Benign", "Malignant"]))
```

```python
plt.figure(figsize=(6, 4))
plt.bar(results["Metric"], results["Value"])
plt.ylim(0, 1.05)
plt.title("Performance Metrics")
plt.ylabel("Score")
plt.show()
```
